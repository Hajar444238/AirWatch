```python
from pathlib import Path
import pandas as pd

gold = Path(r"C:\Users\ADMIN\Desktop\AirWatch\analytics\outputs\aggregated_data\airwatch_gold")
files = sorted((gold / "pm25_sample").glob("part-*.csv"))

if not files:
    raise FileNotFoundError(f"Aucun CSV trouvé dans {gold / 'pm25_sample'}")

df = pd.concat([pd.read_csv(file) for file in files], ignore_index=True)

display(df.head())
df.info()
display(df.describe(include="all"))
```

```python
print(f"Nombre de mesures dans l'échantillon : {len(df):,}")
print(f"Nombre de stations : {df['station'].nunique()}")

station_summary = (
    df.groupby("station", dropna=False)["pm25"]
    .agg(
        observations="count",
        moyenne="mean",
        mediane="median",
        ecart_type="std",
        minimum="min",
        maximum="max",
    )
    .sort_values("moyenne", ascending=False)
    .round(2)
)

display(station_summary)
```
