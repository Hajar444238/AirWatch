```python
from pathlib import Path
import pandas as pd

raw_dir = Path(
    r"C:\Users\ADMIN\Downloads\beijing_air_quality\PRSA_Data_20130301-20170228"
)
csv_files = sorted(raw_dir.glob("PRSA_Data_*.csv"))

if not csv_files:
    raise FileNotFoundError(f"Aucun fichier PRSA_Data_*.csv dans {raw_dir}")

frames = []
for file in csv_files:
    frame = pd.read_csv(file)
    frame["station"] = (
        file.stem
        .replace("PRSA_Data_", "")
        .replace("_20130301-20170228", "")
    )
    frames.append(frame)

data = pd.concat(frames, ignore_index=True)

print(f"Dimensions : {data.shape[0]:,} lignes × {data.shape[1]} colonnes")
print(f"Stations : {data['station'].nunique()}")
display(data.head())
display(data.dtypes.rename("type").to_frame())

print("Valeurs manquantes par colonne :")
display(data.isna().sum().sort_values(ascending=False).to_frame("missing_count"))

data["event_time"] = pd.to_datetime(
    data[["year", "month", "day", "hour"]],
    errors="coerce",
)

print(f"Période : {data['event_time'].min()} — {data['event_time'].max()}")
display(data.groupby("station").size().sort_values(ascending=False).to_frame("observations"))
```
